```python
import torch
inputs = torch.tensor(
  [[0.43, 0.15, 0.89], # Your     (x^1)
   [0.55, 0.87, 0.66], # journey  (x^2)
   [0.57, 0.85, 0.64], # starts   (x^3)
   [0.22, 0.58, 0.33], # with     (x^4)
   [0.77, 0.25, 0.10], # one      (x^5)
   [0.05, 0.80, 0.55]] # step     (x^6)
)

query = inputs[1]
attn_scores_2 = torch.empty(inputs.shape[0])
print(f"attn_scores_2: {attn_scores_2}")
for i, x_i in enumerate(inputs):
    attn_scores_2[i] = torch.dot(x_i, query)
print(attn_scores_2)

attn_scores = torch.empty([6,6])
print(f"\nattn_scores: {attn_scores}")

for i,input_i in enumerate(inputs):
    for j, input_j in enumerate(inputs):
        attn_scores[i][j] = torch.dot(input_i, input_j)


attn_weights = torch.softmax(attn_scores, dim=1)
print(f"\nattention weights : {attn_weights}")
print(f"\nattn_scores: {attn_scores}")

print(attn_weights.sum(dim = 1))

context_vectors = torch.zeros([6,3])


for i in range(inputs.shape[0]):
    for j, attn_w in enumerate(attn_weights[i]):
        for k, input_k in enumerate(inputs[j]):
            context_vectors[i][k] += input_k * attn_w

print(f"\ncontext_vecotrs : {context_vectors}")

```
