```python
import sys
sys.path.append("..") 
import urllib.request
import re
from src.SimpleTokenizer import SimpleTokenizer
url = "https://gutenberg.org/cache/epub/345/pg345.txt"

with urllib.request.urlopen(url) as response:
    text = response.read().decode("utf-8")

start_pat = r"\*\*\* START OF TH(IS|E) PROJECT GUTENBERG EBOOK.*?\*\*\*"
end_pat   = r"\*\*\* END OF TH(IS|E) PROJECT GUTENBERG EBOOK.*?\*\*\*"

start = re.search(start_pat, text)
end   = re.search(end_pat, text)
text = text[start.end():end.start()].strip()


print(f"{text[:500]}\n")

tokens_raw = re.split(r'([,.:;?_!"()\'“”‘’]|--|\s)', text)
print(f"{tokens_raw[:20]}")
print(f"Total tokens count : {len(tokens_raw)}\n")

token_set = set(tokens_raw)
print(token_set)
print(f"Total distinct tokens count : {len(token_set)}\n")


preprocessed = sorted([item.strip() for item in token_set if item.strip()])
preprocessed = preprocessed + ["<|endoftext|>", "<|unk|>"]
print("preprocessed", preprocessed[:200])
print(f"Total distinct preprocessed tokens count : {len(preprocessed)}\n")

vocab = {token:i for i,token in enumerate(preprocessed)}
print(f"vocab : {vocab}\n")

simple_tokenizer = SimpleTokenizer(vocab)

text = "Dublin is the capital of Ireland"
encoded = simple_tokenizer.encode(text)
decoded_text = simple_tokenizer.decode(encoded)

print(f"encoded text: {encoded}")
print(f"decoded text: {decoded_text}")
```

# Basic Tokenization
- Fetched raw text and split it into a token list on whitespace and
  punctuation, keeping the delimiters as tokens (capturing group in re.split)
- deduplicated into a set and stripped whitespace/empty entries, then
  sorted it for stable ordering
- Enumerated the sorted tokens to build the vocab: {token: id} for encoding
  (and {id: token} for decoding)
